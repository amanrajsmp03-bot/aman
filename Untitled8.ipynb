```python
import pandas as pd
import numpy as np
import time

url = "https://raw.githubusercontent.com/shuklaxvaibhav/csa/refs/heads/main/adult.csv"
col_names = ['age', 'workclass', 'fnlwgt', 'education', 'education-num', 'marital-status', 'occupation', 'relationship', 'race', 'sex', 'capital-gain', 'capital-loss', 'hours-per-week', 'native-country', 'income']
df = pd.read_csv(url, names=col_names)
```

```python
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['hours-per-week'] = pd.to_numeric(df['hours-per-week'], errors='coerce')

age_arr = df['age'].to_numpy()
hours_arr = df['hours-per-week'].to_numpy()

print(age_arr[0:5])
print(age_arr[age_arr > 70])
```

```python
start_bc = time.time()
ratio_bc = hours_arr / age_arr
time_bc = time.time() - start_bc

start_loop = time.time()
ratio_loop = [hours_arr[i] / age_arr[i] for i in range(len(age_arr))]
time_loop = time.time() - start_loop

print(time_bc)
print(time_loop)
```

```python
subset = df[(df['age'] > 50) & (df['hours-per-week'] > 40) & (df['income'] == '>50K')]
print(subset.head())
```

```python
grouped = df.groupby('education')['age'].mean().reset_index()

grouped.to_csv('grouped.csv', index=False)
grouped.to_excel('grouped.xlsx', index=False)

df_csv = pd.read_csv('grouped.csv')
df_xlsx = pd.read_excel('grouped.xlsx')

print(df_csv.equals(df_xlsx))
print(df_csv.dtypes.equals(df_xlsx.dtypes))
```
